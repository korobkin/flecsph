# Smoothly truncated power-law shell profile

**FleCSPH — spherical initial-data profile**

We construct a spherical density profile which is a pure power law $r^\alpha$ in a
"body" region and is brought down to zero on both sides by parabolic *rims* of
width $q$, in such a way that the profile is continuous **and** has a continuous
derivative at the two joints.

## 1. Statement of the problem

Let

$$h \;=\; \text{inner hole radius},\qquad q \;=\; \text{rim width},\qquad
\alpha \;=\; \text{power-law index},$$

and define the two joint radii

$$a \equiv h+q, \qquad b \equiv 1-q, \qquad 0 < h < a < b < 1 .$$

We look for

$$\rho(r)=\begin{cases}
0, & r<h\\[2pt]
P_-(r), & h\le r< a\\[2pt]
r^\alpha, & a\le r< b\\[2pt]
P_+(r), & b\le r\le 1\\[2pt]
0, & r>1
\end{cases}$$

where $P_\mp$ are quadratic polynomials subject to

| | inner rim | outer rim |
|---|---|---|
| vanishes at the edge | $P_-(h)=0$ | $P_+(1)=0$ |
| matches the power law | $P_-(a)=f(a)$ | $P_+(b)=f(b)$ |
| **tangent** to the power law | $P_-'(a)=f'(a)$ | $P_+'(b)=f'(b)$ |

with

$$f(r)=r^{\alpha},\qquad f'(r)=\alpha\,r^{\alpha-1}.$$

Each parabola has three free coefficients and is subject to three conditions, so
the solution exists and is **unique**.

## 2. Inner rim, $h\le r\le a=h+q$

It is convenient to expand about the point where the parabola must vanish. Put

$$s \equiv r-h \in[0,q],\qquad P_-(r)=a_-s^2+b_-s+c_- .$$

The condition $P_-(h)=0$ gives immediately $c_-=0$. The remaining two conditions
are imposed at $s=q$ (i.e. $r=a$):

$$a_-q^2+b_-q=f(a), \qquad\qquad 2a_-q+b_-=f'(a).$$

Multiply the second equation by $q$ and subtract the first:

$$a_-q^2 = q f'(a)-f(a) \quad\Longrightarrow\quad
\boxed{\;a_-=\frac{q\,f'(a)-f(a)}{q^{2}}\;}$$

and back-substituting into the first,

$$\boxed{\;b_-=\frac{2f(a)-q\,f'(a)}{q}\;}$$

### Explicit form

With $f(a)=a^\alpha$ and $q f'(a)=\alpha q\,a^{\alpha-1}=a^{\alpha}\,\kappa_-$, where

$$\kappa_-\;\equiv\;\frac{\alpha q}{a}\;=\;\frac{\alpha q}{h+q}$$

is the dimensionless logarithmic slope of the power law at the joint, the
coefficients become

$$a_-=\frac{a^{\alpha}}{q^{2}}\bigl(\kappa_--1\bigr),\qquad
b_-=\frac{a^{\alpha}}{q}\bigl(2-\kappa_-\bigr).$$

Introducing the rim coordinate $u\equiv (r-h)/q\in[0,1]$ we get the compact form

$$\boxed{\;P_-(r)=a^{\alpha}\Bigl\{\,\bigl[1-(1-u)^{2}\bigr]\;-\;\kappa_-\,u\,(1-u)\Bigr\},
\qquad u=\frac{r-h}{q}\;}$$

The first term is exactly the "mesa" rim already used in the code
(`rho_shell_density`), which is recovered for $\alpha=0$ ($\kappa_-=0$); the
second term is the tilt required to make the parabola tangent to $r^\alpha$
rather than to a horizontal plateau.

## 3. Outer rim, $b=1-q\le r\le 1$

Same trick, expanding about the outer zero $r=1$:

$$t\equiv r-1\in[-q,0],\qquad P_+(r)=a_+t^{2}+b_+t \quad(\text{so } P_+(1)=0).$$

The matching conditions are now imposed at $t=-q$ (i.e. $r=b$):

$$a_+q^{2}-b_+q=f(b),\qquad\qquad -2a_+q+b_+=f'(b).$$

Multiplying the second by $q$ and adding to the first,

$$-a_+q^{2}=f(b)+q f'(b)\quad\Longrightarrow\quad
\boxed{\;a_+=-\frac{f(b)+q\,f'(b)}{q^{2}}\;},\qquad
\boxed{\;b_+=-\frac{2f(b)+q\,f'(b)}{q}\;}$$

### Explicit form

With $\kappa_+\equiv \alpha q/b=\alpha q/(1-q)$,

$$a_+=-\frac{b^{\alpha}}{q^{2}}\bigl(1+\kappa_+\bigr),\qquad
b_+=-\frac{b^{\alpha}}{q}\bigl(2+\kappa_+\bigr),$$

and with the outer rim coordinate $v\equiv(1-r)/q\in[0,1]$,

$$\boxed{\;P_+(r)=b^{\alpha}\Bigl\{\,\bigl[1-(1-v)^{2}\bigr]\;+\;\kappa_+\,v\,(1-v)\Bigr\},
\qquad v=\frac{1-r}{q}\;}$$

The two rims are mirror images of each other, the only difference being the sign
in front of $\kappa$: going *outward* through the inner rim the profile must
arrive with slope $+\alpha a^{\alpha-1}$, going *inward* through the outer rim it
must arrive with slope $-\alpha b^{\alpha-1}$ in $v$.

## 4. Summary — the complete profile

$$\rho(r)=\rho_0\times\begin{cases}
0, & r<h,\\[6pt]
(h+q)^{\alpha}\Bigl\{1-(1-u)^{2}-\kappa_-u(1-u)\Bigr\},\quad u=\dfrac{r-h}{q},
 & h\le r<h+q,\\[10pt]
r^{\alpha}, & h+q\le r<1-q,\\[6pt]
(1-q)^{\alpha}\Bigl\{1-(1-v)^{2}+\kappa_+v(1-v)\Bigr\},\quad v=\dfrac{1-r}{q},
 & 1-q\le r\le 1,\\[10pt]
0, & r>1,
\end{cases}$$

$$\kappa_-=\frac{\alpha q}{h+q},\qquad \kappa_+=\frac{\alpha q}{1-q}.$$

Equivalently, in the "coefficient" form convenient for the C++ implementation:

$$P_-(r)=a_-(r-h)^2+b_-(r-h),\qquad P_+(r)=a_+(r-1)^2+b_+(r-1),$$

$$a_-=\frac{qf'(a)-f(a)}{q^{2}},\quad b_-=\frac{2f(a)-qf'(a)}{q},\quad
a_+=-\frac{f(b)+qf'(b)}{q^{2}},\quad b_+=-\frac{2f(b)+qf'(b)}{q}.$$

**Checks.** For $\alpha=0$ the profile reduces to the existing flat-top "mesa"
shell, $\rho=1-\bigl(r-(h+q)\bigr)^2/q^2$ and $\rho=1-\bigl(r-(1-q)\bigr)^2/q^2$.
For $h\to 0$, $q\to 0$ it reduces to the bare power law on $[0,1]$.

## 5. When does the profile stay non-negative?

Both rims are parabolas with a root at the outer edge, so we must check that the
second root does not fall *inside* the rim. Factorising the boxed forms,

$$P_-/a^\alpha = u\bigl[(2-\kappa_-)-(1-\kappa_-)u\bigr],\qquad
P_+/b^\alpha = v\bigl[(2+\kappa_+)-(1+\kappa_+)v\bigr].$$

The bracket is *linear* in $u$ (resp. $v$), and at $u=1$ (resp. $v=1$) it equals
$1>0$; therefore its minimum on $[0,1]$ is attained at the other endpoint, and
non-negativity is equivalent to the value at $u=0$ (resp. $v=0$) being
non-negative:

$$\boxed{\;\kappa_-\le 2\iff \alpha q\le 2(h+q)\;}\qquad\qquad
\boxed{\;\kappa_+\ge -2\iff \alpha q\ge -2(1-q)\;}$$

The first condition bites for steeply **rising** profiles ($\alpha>0$ with a
narrow inner rim), the second for steeply **falling** ones ($\alpha<0$). Both are
sharp: exceeding them by any amount produces a negative lobe inside the rim
(verified numerically below). A practical rule of thumb: keep
$q<2(h+q)/\alpha$ for $\alpha>0$.

The profile is $C^1$ at $r=a$ and $r=b$ by construction. At $r=h$ and $r=1$ it is
only $C^0$: the slope jumps from $0$ to
$P_-'(h)=b_-=a^\alpha(2-\kappa_-)/q$ and from
$P_+'(1)=b_+=-b^\alpha(2+\kappa_+)/q$ to $0$. This is the same (intended)
behaviour as the existing mesa profile.

## 6. Derivative and enclosed mass

Both are needed by `density_profiles.h` (`spherical_drho_dr` and
`spherical_mass_profile`).

### 6.1 Logarithmic derivative

$$\frac{d\rho}{dr}=\begin{cases}
0,& r<h,\\
2a_-(r-h)+b_-, & h\le r<a,\\
\alpha r^{\alpha-1}, & a\le r<b,\\
2a_+(r-1)+b_+, & b\le r\le 1,\\
0,& r>1.
\end{cases}$$

### 6.2 Enclosed mass

$M(r)=4\pi\int_0^r r'^2\rho(r')\,dr'$. In the inner rim substitute $s=r-h$,
$r^2=s^2+2hs+h^2$, so that

$$r^{2}P_-=a_-s^{4}+(b_-+2ha_-)s^{3}+(2hb_-+h^{2}a_-)s^{2}+h^{2}b_-s,$$

and, since $M(h)=0$, the integration constant vanishes identically:

$$\boxed{\;M_-(r)=4\pi\Bigl[\tfrac{a_-}{5}s^{5}+\tfrac{b_-+2ha_-}{4}s^{4}
+\tfrac{2hb_-+h^{2}a_-}{3}s^{3}+\tfrac{h^{2}b_-}{2}s^{2}\Bigr]_{s=r-h}\;}$$

In the body region ($\alpha\ne-3$),

$$\boxed{\;M(r)=M(a)+\frac{4\pi}{\alpha+3}\bigl(r^{\alpha+3}-a^{\alpha+3}\bigr)\;}
\qquad
\bigl(\alpha=-3:\;M(r)=M(a)+4\pi\ln(r/a)\bigr)$$

In the outer rim, with $t=r-1$ and
$r^{2}P_+=a_+t^{4}+(b_++2a_+)t^{3}+(2b_++a_+)t^{2}+b_+t$,

$$G(t)\equiv\tfrac{a_+}{5}t^{5}+\tfrac{b_++2a_+}{4}t^{4}
+\tfrac{2b_++a_+}{3}t^{3}+\tfrac{b_+}{2}t^{2},
\qquad
\boxed{\;M_+(r)=M(b)+4\pi\bigl[G(r-1)-G(-q)\bigr]\;}$$

so that the total mass is $M_{\rm tot}=M(b)-4\pi G(-q)$ (since $G(0)=0$).
Setting $\rho_0=1/M_{\rm tot}$ normalises the shell to unit total mass.

> **Note.** Each piece is written so that the integration constant is fixed by
> the *lower* limit of that piece — this is exactly what the current
> `mass_shell_density()` in `include/physics/density_profiles.h` is missing, and
> why it returns a non-zero (negative) value at $r=h$.

## 7. Implementation and verification

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

FOURPI = 4.0*np.pi


class ShellProfile:
    """rho(r) = 0 | parabola | r^alpha | parabola | 0, C^1 at both joints."""

    def __init__(self, h=0.3, q=0.1, alpha=0.0, normalize=False):
        assert 0.0 <= h and h + q < 1.0 - q, "need h < h+q < 1-q"
        self.h, self.q, self.alpha = h, q, alpha
        a, b = h + q, 1.0 - q
        self.a, self.b = a, b

        f  = lambda r: r**alpha            # power law
        df = lambda r: alpha*r**(alpha-1)  # and its derivative

        # inner rim: P- = am*(r-h)^2 + bm*(r-h)
        self.am = (q*df(a) - f(a))/q**2
        self.bm = (2.0*f(a) - q*df(a))/q
        # outer rim: P+ = ap*(r-1)^2 + bp*(r-1)
        self.ap = -(f(b) + q*df(b))/q**2
        self.bp = -(2.0*f(b) + q*df(b))/q

        # mass of the inner rim and of the body, used as integration constants
        self.M_a = FOURPI*self._F(q)
        if abs(alpha + 3.0) > 1e-12:
            self.M_b = self.M_a + FOURPI*(b**(alpha+3) - a**(alpha+3))/(alpha+3)
        else:
            self.M_b = self.M_a + FOURPI*np.log(b/a)
        self.M_tot = self.M_b - FOURPI*self._G(-q)

        self.rho0 = 1.0/self.M_tot if normalize else 1.0

    # antiderivatives of r^2 * P_-(r) and r^2 * P_+(r)
    def _F(self, s):
        h, am, bm = self.h, self.am, self.bm
        return (am/5*s**5 + (bm + 2*h*am)/4*s**4
                + (2*h*bm + h*h*am)/3*s**3 + h*h*bm/2*s**2)

    def _G(self, t):
        ap, bp = self.ap, self.bp
        return (ap/5*t**5 + (bp + 2*ap)/4*t**4
                + (2*bp + ap)/3*t**3 + bp/2*t**2)

    def _masks(self, r):
        h, a, b = self.h, self.a, self.b
        return ((r >= h) & (r < a), (r >= a) & (r < b),
                (r >= b) & (r <= 1.0), r > 1.0)

    def rho(self, r):
        r = np.atleast_1d(np.asarray(r, float))
        out = np.zeros_like(r)
        m1, m2, m3, _ = self._masks(r)
        s = r[m1] - self.h
        out[m1] = self.am*s*s + self.bm*s
        out[m2] = r[m2]**self.alpha
        t = r[m3] - 1.0
        out[m3] = self.ap*t*t + self.bp*t
        return self.rho0*out

    def drho_dr(self, r):
        r = np.atleast_1d(np.asarray(r, float))
        out = np.zeros_like(r)
        m1, m2, m3, _ = self._masks(r)
        out[m1] = 2*self.am*(r[m1] - self.h) + self.bm
        out[m2] = self.alpha*r[m2]**(self.alpha - 1.0)
        out[m3] = 2*self.ap*(r[m3] - 1.0) + self.bp
        return self.rho0*out

    def mass(self, r):
        r = np.atleast_1d(np.asarray(r, float))
        out = np.zeros_like(r)
        m1, m2, m3, m4 = self._masks(r)
        al = self.alpha
        out[m1] = FOURPI*self._F(r[m1] - self.h)
        if abs(al + 3.0) > 1e-12:
            out[m2] = self.M_a + FOURPI*(r[m2]**(al+3) - self.a**(al+3))/(al+3)
        else:
            out[m2] = self.M_a + FOURPI*np.log(r[m2]/self.a)
        out[m3] = self.M_b + FOURPI*(self._G(r[m3] - 1.0) - self._G(-self.q))
        out[m4] = self.M_tot
        return self.rho0*out

    def positive(self):
        """non-negativity criteria of Sec. 5"""
        km = self.alpha*self.q/self.a
        kp = self.alpha*self.q/self.b
        return (km <= 2.0) and (kp >= -2.0)
````

### 7.1 Checks

For a range of $(h,q,\alpha)$ we verify

* $\rho(h)=\rho(1)=0$;
* continuity **and** tangency at $r=a$ and $r=b$ (jump in $\rho$ and in
  $d\rho/dr$ measured across the joints);
* the analytic $M(r)$ against direct numerical quadrature of
  $4\pi\int r^2\rho\,dr$;
* the non-negativity criterion against the actual minimum of $\rho$.

````python
eps = 1e-7
cases = [(0.3, 0.10,  0.0), (0.3, 0.10, -1.5), (0.3, 0.10,  2.0),
         (0.2, 0.05, -3.0), (0.5, 0.20,  1.0), (0.1, 0.30, -2.0),
         (0.05, 0.10, 3.1)]

hdr = f"{'h':>5}{'q':>6}{'alpha':>7} | {'rho(h)':>9}{'rho(1)':>9} | " \
      f"{'[rho]_a':>9}{'[rho.]_a':>9}{'[rho]_b':>9}{'[rho.]_b':>9} | " \
      f"{'massErr':>9} | {'min rho':>10}  pos?"
print(hdr); print('-'*len(hdr))
for h, q, al in cases:
    P = ShellProfile(h, q, al)
    a, b = P.a, P.b
    jr_a = P.rho(a+eps)[0] - P.rho(a-eps)[0]
    jd_a = P.drho_dr(a+eps)[0] - P.drho_dr(a-eps)[0]
    jr_b = P.rho(b+eps)[0] - P.rho(b-eps)[0]
    jd_b = P.drho_dr(b+eps)[0] - P.drho_dr(b-eps)[0]
    err = max(abs(quad(lambda x: FOURPI*x*x*P.rho(x)[0], 0, rr,
                       points=[h, a, b], limit=400)[0] - P.mass(rr)[0])
              for rr in np.linspace(h, 1.0, 11))
    rmin = P.rho(np.linspace(h, 1.0, 20001)).min()
    print(f"{h:5.2f}{q:6.2f}{al:7.2f} | {P.rho(h)[0]:9.1e}{P.rho(1.0)[0]:9.1e} | "
          f"{jr_a:9.1e}{jd_a:9.1e}{jr_b:9.1e}{jd_b:9.1e} | {err:9.1e} | "
          f"{rmin:10.2e}  {str(P.positive()):>5}")
````

All jumps and the mass error are at round-off level, and the last line shows
the sharpness of the criterion of Sec. 5: $h=0.05,\,q=0.1,\,\alpha=3.1$ violates
$\alpha q\le 2(h+q)$ (i.e. $0.31>0.30$) and indeed develops a small negative
lobe.

### 7.2 The profile, its derivative and the enclosed mass

````python
h, q = 0.3, 0.1
alphas = [2.0, 1.0, 0.0, -1.0, -2.0]
colors = plt.cm.viridis(np.linspace(0.05, 0.85, len(alphas)))
r = np.linspace(0.0, 1.15, 4000)

# all profiles normalised to unit total mass, so that they are comparable
fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
for al, col in zip(alphas, colors):
    P = ShellProfile(h, q, al, normalize=True)
    ax[0].plot(r, P.rho(r),     color=col, lw=1.8, label=rf'$\alpha={al:+.0f}$')
    ax[1].plot(r, P.drho_dr(r), color=col, lw=1.8)
    ax[2].plot(r, P.mass(r),    color=col, lw=1.8)
    # the bare power law the rims are tangent to, continued through the rims
    rr = np.linspace(h, 1.0, 400)
    ax[0].plot(rr, P.rho0*rr**al, color=col, lw=0.8, ls=':')

for k, (ttl, ylab) in enumerate([('density', r'$\rho(r)$'),
                                 (r'derivative (symlog)', r'$d\rho/dr$'),
                                 ('enclosed mass', r'$M(r)$')]):
    ax[k].axvspan(h, h+q, color='0.85', zorder=0)     # inner rim
    ax[k].axvspan(1-q, 1.0, color='0.85', zorder=0)   # outer rim
    for x in (h, h+q, 1-q, 1.0):
        ax[k].axvline(x, color='0.6', lw=0.6, ls='--', zorder=0)
    ax[k].set_xlabel('$r$'); ax[k].set_ylabel(ylab); ax[k].set_title(ttl)
    ax[k].set_xlim(0, 1.15)
ax[0].set_ylim(-0.05, 1.0)
ax[1].set_yscale('symlog', linthresh=1.0)
ax[1].set_ylim(-30, 30)
ax[2].set_ylim(-0.02, 1.05)
ax[0].legend(fontsize=9, loc='upper left')
for k in (0, 1, 2):
    ax[k].text(h+q/2, 0.96, 'rim', ha='center', fontsize=8, color='0.4',
               transform=ax[k].get_xaxis_transform())
    ax[k].text(1-q/2, 0.96, 'rim', ha='center', fontsize=8, color='0.4',
               transform=ax[k].get_xaxis_transform())
fig.suptitle(rf'shell profile: $h={h}$, $q={q}$, unit total mass   '
             r'(dotted: the bare power law $\propto r^\alpha$)', y=1.02)
fig.tight_layout()
plt.show()
````

The rims (shaded) join the power law with no kink: $d\rho/dr$ (middle panel)
is continuous at $r=h+q$ and $r=1-q$, and jumps only at the outer edges $r=h$ and
$r=1$, where the profile hits zero. The dotted curves show the power law the rims
are tangent to.

### 7.3 Approaching the non-negativity limit

````python
h, q = 0.05, 0.1
a = h + q
al_crit = 2.0*a/q          # alpha*q = 2(h+q)
r = np.linspace(0, 1.02, 3000)

fig, ax = plt.subplots(figsize=(7.2, 4.2))
for fac, ls in [(0.5, '-'), (0.9, '-'), (1.0, '--'), (1.15, ':')]:
    al = fac*al_crit
    P = ShellProfile(h, q, al)
    ax.plot(r, P.rho(r)/P.rho(a)[0], ls, lw=1.8,
            label=rf'$\alpha={al:.2f}={fac:g}\,\alpha_{{\rm crit}}$'
                  + ('  (negative!)' if not P.positive() else ''))
ax.axhline(0, color='k', lw=0.7)
ax.axvspan(h, a, color='0.88', zorder=0)
ax.set_xlim(0, 0.25); ax.set_ylim(-0.15, 1.3)
ax.set_xlabel('$r$'); ax.set_ylabel(r'$\rho/\rho(h+q)$')
ax.set_title(rf'inner rim, $h={h}$, $q={q}$: '
             rf'$\alpha_{{\rm crit}}=2(h+q)/q={al_crit:g}$')
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
````

At $\alpha=\alpha_{\rm crit}=2(h+q)/q$ the inner rim becomes tangent to zero
*at the hole edge* ($P_-'(h)=b_-=0$), and beyond it the parabola dips negative.
The mirror condition at the outer rim, $\alpha\ge-2(1-q)/q$, is much weaker for
typical parameters ($h\ll1$).

## 8. Translation to C++

For `include/physics/density_profiles.h` the coefficients need to be computed
once (in the profile-selection block, next to `mesa_q`, `mesa_hole`,
`mesa_rho0`), for instance

```cpp
  const double a = shell_inner_hole + mesa_q,  b = 1. - mesa_q;
  const double fa = pow(a, shell_alpha),  dfa = shell_alpha*pow(a, shell_alpha-1.);
  const double fb = pow(b, shell_alpha),  dfb = shell_alpha*pow(b, shell_alpha-1.);
  shell_am = (mesa_q*dfa - fa)/SQ(mesa_q);
  shell_bm = (2.*fa - mesa_q*dfa)/mesa_q;
  shell_ap = -(fb + mesa_q*dfb)/SQ(mesa_q);
  shell_bp = -(2.*fb + mesa_q*dfb)/mesa_q;
```

after which `rho`, `drho/dr` and the mass profile are the five-branch
expressions of Secs. 4 and 6, with the integration constants
$M(a)$ and $M(b)$ also precomputed. Setting
$\rho_0=1/M_{\rm tot}$ normalises the shell to unit mass; the $\alpha=0$ case
must reproduce the present `shell` profile exactly, which is a useful
regression test.
